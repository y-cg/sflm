```python
from torch.utils.data import DataLoader

from sflm.dataset import LangABC
from sflm.utils import TokenizerABC
```

```python
str_len = 10
```

```python
tokenizer = TokenizerABC()
tokenize = lambda x: tokenizer.encode(x, pad_to_length=str_len + 2)
```

```python
dataset = LangABC(size=100, max_len=str_len, transform=tokenize)
```

```python
for i in range(5):
    print(dataset[i])
```

```python
dataloader = DataLoader(dataset, batch_size=8)
```

```python
batcher = iter(dataloader)
```

```python
batch = next(batcher)
```
